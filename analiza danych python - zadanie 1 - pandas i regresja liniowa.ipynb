# Analiza danych: Edukacyjnej Sieci Antysmogowej (ESA)

1.  Zapoznaj się ze  [źródłem danych](https://dane.gov.pl/pl/dataset/2913,dane-pomiarowe-esa-edukacyjna-siec-antysmogowa) na portalu dane.gov.pl i na  [stronie projektu](https://esa.nask.pl/) Edukacyjnej Sieci Antysmogowej (ESA)
1.  Zaczytaj dane ESA z linku: [dane](https://public-esa.ose.gov.pl/api/v1/smog). Sprawdź jakość danych (braki danych, błędne wartości) i usuń problemy w danych - samodzielnie zaproponuj rozwiązanie.
1. Wybierz 5% szkół z największym stężeniem:
 - PM10_AVG
 - PM25_AVG  
Policz jaki procent szkół jest jednocześnie w grupie z największym stężeniem PM10_AVG i PM25_AVG.
1. Policz wartość średnią oraz odchylenie standardowe dla
 - HUMIDITY_AVG
 - PRESSURE_AVG
 - TEMPERATURE_AVG
 - PM10_AVG
 - PM25_AVG  
w takich grupach:  
 - POST_CODE
 - CITY
 - [OKRĘGI KODU POCZTOWEGO](https://pl.wikipedia.org/wiki/Kody_pocztowe_w_Polsce) (na podstawie pierwszego znaku kodu pocztowego)
Opisz wnioski i zapisz wyniki w pliku excela.

7. Utwórz mapę korzystając z np z biblioteki `folium`. Pokaż na mapie lokalizacje dla jakości powietrza i wartości dla `pm10_avg`.

8. Zbuduj model uczenia maszynowego korzystając z regresji liniowej, który będzie przewidywał wartości stężenia zmiennej `pm25_avg` (stężenie pyłu zawieszonego 2.5).
