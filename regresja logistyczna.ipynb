```python
import pandas as pd
```

```python
df = pd.read_csv('survey lung cancer.csv')
```

```python
df.shape
```

```python
df.to_excel('lung_cancer.xlsx')
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=['GENDER'], drop_first=True)
```

```python
df
```

```python
X, y = df.drop('LUNG_CANCER', axis=1), df['LUNG_CANCER']
```

```python
X
```

```python
from sklearn import linear_model
```

```python
clf  = linear_model.LogisticRegression(max_iter=1000)
```

```python
clf.fit(X, y)
```

```python
clf.score(X, y)  # co to jest score dla modeli klasyfikacji
```

```python
clf.predict(X)
```

```python
df['LUNG_CANCER']
```

```python
(clf.predict(X) == df['LUNG_CANCER']).mean()
```

# podział na train i na test

```python
from sklearn import model_selection
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.4, random_state=4)
```

```python
clf.fit(X_train, y_train)
```

```python
clf.score(X_train, y_train), clf.score(X_test, y_test)
```
